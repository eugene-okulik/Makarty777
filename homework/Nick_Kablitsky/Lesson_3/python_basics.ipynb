Задачи по обработке чисел:

Даны 2 числа a и b. Получить их сумму, разность и произведение
Даны числа x и y. Получить x − y / 5 + xy
Даны два числа. Найти среднее арифметическое и среднее геометрическое этих чисел

```python
# Даны 2 числа a и b. Получить их сумму, разность и произведение
a_int , b_int = int(input("Введите число a : ")), int(input("Введите число b : "))
sum_ab, del_ab, compos_ab = (a_int + b_int), (a_int - b_int), (a_int * b_int)  
```

```python
# Даны числа x и y. Получить x − y / 5 + xy
x_int, y_int = int(input("Введите число x : ")), int(input("Введите число y : "))
match_xy = (x_int - y_int)/(5 + (x_int * y_int))
```

```python
# Даны два числа. Найти среднее арифметическое и среднее геометрическое этих чисел
var1_int, var2_int = int(input("Введите перове число: ")), int(input("Введите второе число: "))
MidleArif, MidleGeom = ((var1_int + var2_int) / 2), ((var1_int * var2_int) ** 2)
```

```python
# Даны катеты прямоугольного треугольника. Найти его гипотенузу и площадь
katet1_int, katet2_int = int(input("Введите первый катет: ")), int(input("Введите второй катет: "))
giptenza_katet1_katet2, square = ((katet1_int + katet2_int) ** 2), ((katet1_int * katet2_int) * 0,5) 
```
